## Pengertian Analisis Data

Analisis data adalah proses pemeriksaan, pembersihan, transformasi, dan pemodelan data dengan tujuan menemukan informasi yang berguna, mendukung kesimpulan, dan membantu pengambilan keputusan. Dalam konteks bisnis dan organisasi, analisis data menjadi sangat penting karena beberapa alasan:

*   **Mengidentifikasi Tren dan Pola:** Analisis data membantu mengidentifikasi tren, pola, dan hubungan yang mungkin tidak terlihat dalam data mentah. Ini memungkinkan organisasi untuk memahami apa yang terjadi, mengapa itu terjadi, dan memprediksi apa yang mungkin terjadi di masa depan.
*   **Pengambilan Keputusan Berbasis Bukti:** Daripada membuat keputusan berdasarkan intuisi atau asumsi, analisis data menyediakan bukti konkret. Keputusan yang didukung data cenderung lebih akurat, efektif, dan menghasilkan hasil yang lebih baik.
*   **Optimalisasi Proses Bisnis:** Dengan menganalisis data operasional, perusahaan dapat mengidentifikasi inefisiensi, bottleneck, dan area untuk perbaikan, yang mengarah pada peningkatan efisiensi dan pengurangan biaya.
*   **Personalisasi Pengalaman Pelanggan:** Data pelanggan dapat dianalisis untuk memahami preferensi, perilaku, dan kebutuhan mereka, memungkinkan perusahaan untuk menawarkan produk, layanan, dan pengalaman yang lebih personal dan relevan.
*   **Mitigasi Risiko:** Analisis data dapat membantu mendeteksi anomali atau potensi risiko (misalnya, penipuan, kegagalan sistem) lebih awal, sehingga tindakan pencegahan dapat diambil.

Secara keseluruhan, analisis data berkontribusi pada pengambilan keputusan yang didorong oleh data dengan mengubah data mentah menjadi wawasan yang dapat ditindaklanjuti. Ini memungkinkan organisasi untuk lebih responsif terhadap perubahan pasar, meningkatkan daya saing, dan mencapai tujuan strategis mereka.

## Perbedaan Descriptive Analysis dan EDA

Descriptive Analysis (Analisis Deskriptif) dan Exploratory Data Analysis (EDA) adalah dua fase penting dalam proses analisis data, meskipun seringkali tumpang tindih dan saling melengkapi. Berikut adalah perbedaan utamanya:

### 1. Descriptive Analysis (Analisis Deskriptif)

*   **Tujuan:** Untuk meringkas dan mendeskripsikan fitur utama dari kumpulan data secara kuantitatif. Ini berfokus pada apa yang telah terjadi di masa lalu. Tujuannya adalah untuk memberikan gambaran yang jelas dan ringkas tentang data.
*   **Metodologi:** Melibatkan perhitungan statistik ringkasan seperti mean, median, mode, standar deviasi, varians, frekuensi, persentase, serta pembuatan tabel distribusi frekuensi dan grafik dasar (misalnya, histogram, bar chart) untuk menunjukkan karakteristik data.
*   **Jenis Pertanyaan:** Menjawab pertanyaan "Apa yang terjadi?" atau "Bagaimana data ini terlihat?". Contoh: "Berapa rata-rata penyewaan sepeda per hari?", "Berapa proporsi penyewaan di musim panas?", "Apa distribusi usia pengguna?"
*   **Output:** Ringkasan statistik, tabel, dan grafik sederhana yang menyajikan fakta-fakta tentang data.

### 2. Exploratory Data Analysis (EDA)

*   **Tujuan:** Untuk menemukan pola, mendeteksi anomali, menguji hipotesis, dan memeriksa asumsi dengan ringkasan grafis dan visualisasi data. EDA lebih berfokus pada pemahaman data secara mendalam sebelum pemodelan formal.
*   **Metodologi:** Melibatkan teknik visualisasi yang lebih canggih (misalnya, scatter plot, box plot, heatmaps, time series plots), pengelompokan (clustering), reduksi dimensi, dan analisis korelasi. Seringkali bersifat iteratif dan fleksibel, memungkinkan analis untuk 'berinteraksi' dengan data.
*   **Jenis Pertanyaan:** Menjawab pertanyaan "Mengapa ini terjadi?", "Hubungan apa yang ada di antara variabel-variabel?", "Apakah ada outlier atau tren yang tidak biasa?". Contoh: "Bagaimana suhu memengaruhi jumlah penyewaan sepeda?", "Apakah ada pola musiman dalam permintaan sepeda?", "Apakah ada perbedaan penggunaan sepeda antara hari kerja dan akhir pekan?"
*   **Output:** Insight tentang hubungan antar variabel, identifikasi masalah data (missing values, outliers), pola yang tersembunyi, dan pemahaman yang lebih dalam yang akan memandu pemilihan model atau analisis lebih lanjut.

### Ringkasan Perbedaan:

| Fitur            | Descriptive Analysis                         | Exploratory Data Analysis (EDA)                          |
| :--------------- | :------------------------------------------- | :------------------------------------------------------- |
| **Tujuan**       | Meringkas & mendeskripsikan data             | Mengungkap pola, hubungan, anomali, hipotesis            |
| **Fokus**        | Apa yang sudah terjadi                       | Mengapa itu terjadi, apa yang mungkin terjadi            |
| **Metodologi**   | Statistik ringkasan, grafik dasar            | Visualisasi interaktif, korelasi, deteksi anomali        |
| **Output**       | Fakta & angka ringkasan                      | Insight, pemahaman data mendalam, panduan pemodelan      |
| **Sifat**        | Konklusif, menjelaskan                       | Penemuan, menyelidiki, menguji asumsi                    |

Secara sederhana, Descriptive Analysis memberitahu kita **apa** yang ada dalam data, sedangkan EDA membantu kita memahami **mengapa** itu ada di sana dan **apa implikasinya**.

## Pentingnya Data Cleaning dan Wrangling

Data cleaning (pembersihan data) dan data wrangling (penataan data) adalah langkah-langkah krusial dalam setiap proyek analisis data. Meskipun seringkali memakan waktu, proses ini memastikan bahwa data yang digunakan untuk analisis akurat, konsisten, dan siap untuk diinterpretasikan. Mengabaikan tahapan ini dapat menyebabkan hasil analisis yang bias, tidak akurat, atau menyesatkan.

Berikut adalah beberapa alasan mengapa data cleaning dan data wrangling sangat penting:

1.  **Menangani Missing Values (Nilai yang Hilang):** Data seringkali memiliki 'lubang' di dalamnya. Nilai yang hilang dapat menyebabkan kesalahan dalam perhitungan statistik, model yang bias, atau bahkan kegagalan dalam proses analisis. Penanganan yang tepat (misalnya, mengisi dengan median/mean, interpolasi, atau menghapus baris/kolom) sangat penting.

2.  **Mengoreksi Tipe Data:** Kesalahan tipe data (misalnya, tanggal disimpan sebagai string, angka sebagai objek) dapat menghambat operasi matematika atau analisis berbasis waktu. Mengonversi tipe data ke format yang benar (misalnya, `object` ke `datetime`, `int` ke `category`) memungkinkan pemanfaatan fitur-fitur pustaka analisis data secara efektif.

3.  **Menghilangkan Duplikat:** Baris data yang duplikat dapat menggelembungkan ukuran dataset secara artifisial dan menyebabkan overestimasi atau bias dalam hasil analisis. Menghapus entri duplikat memastikan bahwa setiap observasi unik dipertimbangkan hanya sekali.

4.  **Mengelola Outlier (Pencilan):** Outlier adalah nilai data yang secara signifikan berbeda dari mayoritas data. Mereka bisa jadi merupakan kesalahan input, anomali yang valid, atau indikasi peristiwa langka. Tergantung pada konteksnya, outlier dapat sangat mempengaruhi hasil analisis statistik dan performa model. Identifikasi dan penanganan yang tepat (misalnya, penghapusan, transformasi, atau imputasi) diperlukan.

5.  **Standardisasi dan Normalisasi:** Untuk data yang berasal dari berbagai sumber atau memiliki skala yang berbeda, standardisasi atau normalisasi (misalnya, untuk fitur cuaca seperti suhu, kelembaban, dan kecepatan angin) memastikan bahwa tidak ada fitur yang mendominasi yang lain secara tidak proporsional dalam analisis atau model.

6.  **Membuat Data Konsisten:** Data yang tidak konsisten (misalnya, format penulisan yang berbeda untuk kategori yang sama, unit pengukuran yang bervariasi) perlu diseragamkan agar analisis dapat dilakukan dengan benar.

Dengan melakukan data cleaning dan wrangling secara menyeluruh, kita dapat memastikan bahwa analisis yang dilakukan akan menghasilkan insight yang lebih andal, valid, dan dapat ditindaklanjuti untuk mendukung pengambilan keputusan bisnis.

## Penjelasan Teknik Machine Learning yang Digunakan

### Teknik Machine Learning: Regresi Linear

Dalam analisis ini, kami menggunakan model **Regresi Linear (Linear Regression)**. Regresi Linear adalah salah satu algoritma Machine Learning paling dasar dan umum yang digunakan untuk memprediksi nilai numerik kontinu.

Secara sederhana, model Regresi Linear berusaha menemukan hubungan linear terbaik antara satu atau lebih variabel independen (fitur) dengan variabel dependen (target). Hubungan ini direpresentasikan dalam bentuk persamaan garis lurus (atau hyperplane dalam dimensi lebih tinggi):

`Y = β0 + β1X1 + β2X2 + ... + βnXn + ε`

Dimana:
*   `Y` adalah variabel dependen (target, dalam kasus ini `cnt` atau jumlah total penyewaan sepeda).
*   `X1, X2, ..., Xn` adalah variabel independen (fitur, seperti `temp`, `hum`, `season`, `yr`, dll.).
*   `β0` adalah *intercept* (nilai Y ketika semua X adalah nol).
*   `β1, β2, ..., βn` adalah koefisien regresi, yang menunjukkan seberapa besar perubahan `Y` untuk setiap perubahan satu unit pada `X` yang sesuai, dengan asumsi variabel lain tetap konstan.
*   `ε` adalah *error term* (kesalahan).

### Mengapa Regresi Linear Digunakan?

Ada beberapa alasan utama mengapa Regresi Linear dipilih untuk analisis ini, sesuai dengan ketentuan yang diberikan:

1.  **Tujuan Pendukung Insight, Bukan Optimasi Kompleks:**
    *   Salah satu tujuan utama adalah menggunakan Machine Learning sebagai pengayaan untuk mendukung dan menguantifikasi insight yang diperoleh dari Analisis Data Eksplorasi (EDA), bukan untuk membangun model prediksi dengan akurasi tertinggi. Regresi Linear sangat cocok untuk tujuan ini karena kesederhanaannya dan kemampuannya untuk menjelaskan hubungan antar variabel secara langsung.

2.  **Variabel Target Berupa Numerik Kontinu (Regresi):**
    *   Variabel yang ingin diprediksi (`cnt`) adalah jumlah total penyewaan sepeda, yang merupakan nilai numerik kontinu. Regresi Linear dirancang khusus untuk masalah regresi semacam ini.

3.  **Interpretasi yang Mudah:**
    *   Koefisien (coefficient) yang dihasilkan oleh model Regresi Linear sangat mudah diinterpretasikan. Setiap koefisien menunjukkan dampak langsung dari fitur terkait terhadap variabel target. Misalnya, koefisien positif menunjukkan bahwa peningkatan fitur akan meningkatkan target, dan sebaliknya untuk koefisien negatif. Hal ini sangat berharga untuk mendapatkan pemahaman bisnis yang jelas tentang faktor-faktor pendorong permintaan sepeda.

4.  **Baseline Model yang Kuat:**
    *   Regresi Linear seringkali menjadi model baseline yang baik. Meskipun mungkin tidak seakurat model yang lebih kompleks, kinerjanya yang solid (dengan R-squared 0.84 dalam kasus ini) menunjukkan bahwa hubungan linear antara fitur dan target cukup kuat. Ini memberikan fondasi yang baik untuk analisis lebih lanjut dan jika diperlukan, dapat menjadi titik awal untuk eksperimen dengan model yang lebih canggih di masa depan.

5.  **Menghindari Overfitting untuk Insight:**
    *   Karena tujuan utama adalah insight, bukan prediksi yang sangat optimal, model yang lebih sederhana seperti Regresi Linear cenderung kurang rentan terhadap *overfitting* pada data training dibandingkan model yang lebih kompleks, yang dapat menghasilkan interpretasi yang lebih umum dan dapat diandalkan untuk populasi data yang lebih luas.

```python
print("--- Cleaning Data ---")

# 1. Convert 'dteday' to datetime objects
df_day['dteday'] = pd.to_datetime(df_day['dteday'])
df_hour['dteday'] = pd.to_datetime(df_hour['dteday'])
print("Converted 'dteday' to datetime type.")

# 2. Define categorical columns for both dataframes
categorical_cols_common = ['season', 'yr', 'mnth', 'holiday', 'weekday', 'workingday', 'weathersit']

# Convert common categorical columns in df_day
for col in categorical_cols_common:
    df_day[col] = df_day[col].astype('category')

# Convert common categorical columns in df_hour
for col in categorical_cols_common:
    df_hour[col] = df_hour[col].astype('category')

# 3. Convert 'hr' column in df_hour to 'category'
df_hour['hr'] = df_hour['hr'].astype('category')
print("Converted relevant integer columns to 'category' type.")

# 4. Handle 'hum' (humidity) anomaly (values of 0.0)
# df_day
hum_zero_day = df_day[df_day['hum'] == 0.0].shape[0]
if hum_zero_day > 0:
    median_hum_day = df_day[df_day['hum'] != 0.0]['hum'].median()
    df_day['hum'].replace(0.0, median_hum_day, inplace=True)
    print(f"Replaced {hum_zero_day} 'hum' values of 0.0 in df_day with median: {median_hum_day:.4f}")
else:
    print("No 'hum' values of 0.0 found in df_day.")

# df_hour
hum_zero_hour = df_hour[df_hour['hum'] == 0.0].shape[0]
if hum_zero_hour > 0:
    median_hum_hour = df_hour[df_hour['hum'] != 0.0]['hum'].median()
    df_hour['hum'].replace(0.0, median_hum_hour, inplace=True)
    print(f"Replaced {hum_zero_hour} 'hum' values of 0.0 in df_hour with median: {median_hum_hour:.4f}")
else:
    print("No 'hum' values of 0.0 found in df_hour.")

# 5. Display .info() and .describe() for verification
print("\n--- df_day Info after cleaning ---")
df_day.info()

print("\n--- df_hour Info after cleaning ---")
df_hour.info()

print("\n--- df_day Descriptive Statistics after cleaning ---")
print(df_day.describe())

print("\n--- df_hour Descriptive Statistics after cleaning ---")
print(df_hour.describe())
```

## Persiapan Data untuk Machine Learning

### Subtask:
Pilih fitur-fitur yang relevan (variabel independen) dan target (variabel dependen) dari DataFrame `df_day` yang telah dibersihkan. Lakukan One-Hot Encoding untuk fitur kategorikal agar sesuai untuk model Machine Learning.


```python
print("--- Preparing data for Machine Learning ---")

# 1. Buat daftar kolom numerik yang akan digunakan sebagai fitur independen
numerical_cols = ['temp', 'hum', 'windspeed']

# 2. Buat daftar kolom kategorikal yang akan digunakan sebagai fitur independen
categorical_cols = ['season', 'yr', 'mnth', 'holiday', 'weekday', 'workingday', 'weathersit']

# 3. Definisikan variabel independen X
X = df_day[numerical_cols + categorical_cols].copy()
print(f"Initial X shape: {X.shape}")

# 4. Definisikan variabel dependen y
y = df_day['cnt'].copy()
print(f"Initial y shape: {y.shape}")

# 5. Lakukan One-Hot Encoding pada kolom kategorikal di X
X_encoded = pd.get_dummies(X, columns=categorical_cols, drop_first=True, dtype=int)
print("Performed One-Hot Encoding on categorical columns.")

# 6. Tampilkan lima baris pertama dari DataFrame X yang telah di-encode
print("\n--- X (Independent Variables) after One-Hot Encoding ---")
print(X_encoded.head())

# 7. Tampilkan jumlah baris dan kolom dari DataFrame X yang telah di-encode
print(f"\nShape of X after One-Hot Encoding: {X_encoded.shape}")
```

## Pembagian Data Training dan Testing

### Subtask:
Bagi dataset yang telah dipersiapkan (`X_encoded` dan `y`) menjadi data training dan data testing dengan rasio 80% training dan 20% testing.


```python
from sklearn.model_selection import train_test_split

print("--- Splitting data into training and testing sets ---")

# Split the dataset into training and testing sets (80% training, 20% testing)
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)

# Print the shapes of the resulting sets to verify the split
print(f"Shape of X_train: {X_train.shape}")
print(f"Shape of X_test: {X_test.shape}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Shape of y_test: {y_test.shape}")
```

## Pelatihan Model Regresi Linear

### Subtask:
Inisialisasi dan latih model regresi linear menggunakan data training yang telah dibagi.


```python
from sklearn.linear_model import LinearRegression

print("--- Training Linear Regression Model ---")

# 1. Instantiate a LinearRegression model object
model = LinearRegression()

# 2. Fit the instantiated model to the training data
model.fit(X_train, y_train)

# 3. Print a confirmation message
print("Linear Regression model has been successfully trained.")
```

## Evaluasi dan Interpretasi Model

### Subtask:
Prediksi menggunakan data testing, evaluasi kinerja model menggunakan metrik seperti Mean Squared Error (MSE) dan R-squared (R2), serta interpretasikan koefisien model untuk memahami pengaruh setiap fitur terhadap jumlah penyewaan sepeda.


```python
from sklearn.metrics import mean_squared_error, r2_score

print("--- Evaluating and Interpreting the Linear Regression Model ---")

# 1. Gunakan model regresi linear yang telah dilatih (`model`) untuk membuat prediksi pada data `X_test`.
y_pred = model.predict(X_test)

# 2. Hitung Mean Squared Error (MSE) antara `y_test` dan hasil prediksi (`y_pred`).
mse = mean_squared_error(y_test, y_pred)

# 3. Hitung R-squared (R2) score antara `y_test` dan hasil prediksi (`y_pred`).
r2 = r2_score(y_test, y_pred)

# 4. Cetak nilai MSE dan R2 yang telah dihitung.
print(f"Mean Squared Error (MSE): {mse:.2f}")
print(f"R-squared (R2) Score: {r2:.2f}")

# 5. Ambil koefisien (`model.coef_`) dan intercept (`model.intercept_`) dari model yang telah dilatih.
coefficients = model.coef_
intercept = model.intercept_

# 6. Buat DataFrame dari koefisien model, menghubungkan setiap koefisien dengan nama fitur yang sesuai dari `X_encoded`.
feature_names = X_encoded.columns
coefficients_df = pd.DataFrame({'Feature': feature_names, 'Coefficient': coefficients})

# 7. Cetak DataFrame koefisien dan nilai intercept.
print("\n--- Model Coefficients ---")
print(coefficients_df.sort_values(by='Coefficient', ascending=False))
print(f"\nIntercept: {intercept:.2f}")

print("Model evaluation and interpretation complete.")
```
